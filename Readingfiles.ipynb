```python
import os
import fitz  # pymupdf for PDF
import pandas as pd
import docx
from typing import List
```

```python
def extract_text_from_docx(file_path: str) -> str:
    """Extract text from a DOCX file including tables."""
    doc = docx.Document(file_path)
    text = []

    for para in doc.paragraphs:
        text.append(para.text)

    for table in doc.tables:
        for row in table.rows:
            row_text = [cell.text.strip() for cell in row.cells]
            text.append("\t".join(row_text))  # Tab-separated row

    return "\n".join(text)
```

```python
def extract_text_from_pdf(file_path: str) -> str:
    """Extract text from a PDF file."""
    doc = fitz.open(file_path)
    text = []

    for page in doc:
        text.append(page.get_text("text"))  # Extract normal text

    return "\n".join(text)
```

```python
def extract_text_from_csv(file_path: str) -> str:
    """Extract text from a CSV file."""
    df = pd.read_csv(file_path, dtype=str)  # Read as string to avoid errors
    return df.to_csv(index=False, sep="\t")  # Tab-separated values
```

```python
def extract_text_from_excel(file_path: str) -> str:
    """Extract text from an Excel file."""
    xls = pd.ExcelFile(file_path)
    text = []
    
    for sheet_name in xls.sheet_names:
        df = pd.read_excel(xls, sheet_name=sheet_name, dtype=str)
        text.append(f"--- Sheet: {sheet_name} ---")
        text.append(df.to_csv(index=False, sep="\t"))

    return "\n".join(text)
```

```python
def extract_text(file_path: str, output_folder: str = "extracted_text") -> None:
    """Extract text from a file and save it to a .txt file."""
    ext = os.path.splitext(file_path)[-1].lower()

    # Choose the correct extraction function
    if ext == ".docx":
        text = extract_text_from_docx(file_path)
    elif ext == ".pdf":
        text = extract_text_from_pdf(file_path)
    elif ext == ".csv":
        text = extract_text_from_csv(file_path)
    elif ext in [".xlsx", ".xls", ".xlsm"]:
        text = extract_text_from_excel(file_path)
    else:
        raise ValueError(f"Unsupported file format: {ext}")

    # Ensure output folder exists
    os.makedirs(output_folder, exist_ok=True)

    # Define output file path
    base_name = os.path.splitext(os.path.basename(file_path))[0]
    output_file_path = os.path.join(output_folder, f"{base_name}.txt")

    # Save extracted text
    with open(output_file_path, "w", encoding="utf-8") as f:
        f.write(text)

    print(f"Extracted text saved to: {output_file_path}")
```

```python
def process_folder(folder_path: str, output_folder: str = "extracted_text") -> None:
    """Process all supported files in a folder."""
    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)

        # Skip directories
        if os.path.isdir(file_path):
            continue

        try:
            extract_text(file_path, output_folder)
        except ValueError as e:
            print(f"Skipping {file_name}: {e}")
        except Exception as e:
            print(f"Error processing {file_name}: {e}")
```

```python
folder_path = "Dr.X Files/"  
process_folder(folder_path)
```
